##Mathematical Theories

<h1>Loss Function and Optimization Problem</h1>

If we consider **mₖ** as the center (or representative) of each cluster and _assume_ that all data points are assigned to a cluster by **mₖ**, then the error for assigning a data point **xᵢ** to cluster **k** is the distance:


***disclaimer***

<h1> Are $\mathbf{x}$ and $\mathbf{m}$ Just on the $x$-Axis (Ox Line)?

No — in the context of K-means clustering, both $\mathbf{x}$ and $\mathbf{m}$ are **vectors**, not just scalar values on the $x$-axis.

---

<h1> Specifically:

- $\mathbf{x}_i \in \mathbb{R}^d$ is a **data point vector**.  
  Example: If $d = 2$, then $\mathbf{x}_i = (x_i^{(1)}, x_i^{(2)}) = (x, y)$

- $\mathbf{m}_k \in \mathbb{R}^d$ is the **cluster center vector** for cluster $k$, same dimension as $\mathbf{x}_i$.

---

<h1> Example in 2D:

If your data is in 2D space, then:

- $\mathbf{x}_i = (x_i^{(1)}, x_i^{(2)})$
- $\mathbf{m}_k = (m_k^{(1)}, m_k^{(2)})$

Then the squared Euclidean distance is:

$$
\| \mathbf{x}_i - \mathbf{m}_k \|_2^2 = (x_i^{(1)} - m_k^{(1)})^2 + (x_i^{(2)} - m_k^{(2)})^2
$$

So this measures the **full geometric distance** between a data point and a cluster center in $d$-dimensional space — **not** just along the $x$-axis.



$$
\| \mathbf{x}_i - \mathbf{m}_k \|_2^2
$$

We want this error to be as small as possible, so we aim to minimize the following quantity:

$$
\| \mathbf{x}_i - \mathbf{m}_k \|_2^2
$$

Moreover, since **xᵢ** is assigned to cluster **k**, we have:

$$
y_{ik} = 1, \quad y_{ij} = 0, \quad \forall j \ne k
$$

Then, the expression above can be rewritten as:

$$
y_{ik} \| \mathbf{x}_i - \mathbf{m}_k \|_2^2 = \sum_{j=1}^{K} y_{ij} \| \mathbf{x}_i - \mathbf{m}_j \|_2^2
$$


The total loss for the dataset becomes:

$$
\mathcal{L}(\mathbf{Y}, \mathbf{M}) = \sum_{i=1}^{N} \sum_{j=1}^{K} y_{ij} \| \mathbf{x}_i - \mathbf{m}_j \|_2^2
$$

Where:

- **Y** = [ **y₁**, **y₂**, ..., **yₙ** ] is the label matrix.

In K-means, we assign each data point $x_i$ to one of 𝐾 clusters. Instead of just writing “this point belongs to cluster 2,” we often represent the assignment using a label matrix 𝑌.

image.png

- **M** = [ **m₁**, **m₂**, ..., **mₖ** ] are the cluster centers.

This is the **loss function** in the K-means clustering problem: $\mathcal{L}(\mathbf{Y}, \mathbf{M})$ with the constraint described in equation (1).

To summarize, the optimization problem is:

$$
\mathbf{Y}, \mathbf{M} = \arg \min_{\mathbf{Y}, \mathbf{M}} \sum_{i=1}^{N} \sum_{j=1}^{K} y_{ij} \| \mathbf{x}_i - \mathbf{m}_j \|_2^2 \tag{2}
$$

subject to:

$$
y_{ij} \in \{0, 1\} \quad \forall i, j; \quad \sum_{j=1}^{K} y_{ij} = 1 \quad \forall i
$$


<h1> Found $Y$, find $M$ </h1>
Thus, for instance, when we have found all of the centers, which the M has been found, to find y, or find all the label vectors to let the loss function to reach its minimum solution

$$
\mathbf{y_i} = \arg \min_{\mathbf{{y_j}}} \sum_{i=1}^{N} \sum_{j=1}^{K} y_{ij} \| \mathbf{x}_i - \mathbf{m}_j \|_2^2 \tag{2}
$$

<h1>Found $M$, Find $Y$</h1>
when we have found the cluster for each points, find the new center for each clusters to make the loss function to be the smallest.

$$
m_j = \arg \min_{{m_j}} \sum_{t=1}^{N} \| \mathbf{x}_i - \mathbf{m}_j \|_2^2 \tag{2}
$$

$$
\frac{\partial l(\mathbf{m}_j)}{\partial \mathbf{m}_j} = 2 \sum_{i=1}^{N} y_{ij}(\mathbf{m}_j - \mathbf{x}_i)
$$


$$
=> \mathbf{m}_j \sum_{i=1}^{N} y_{ij} = \sum_{i=1}^{N} y_{ij} \mathbf{x}_i
$$

$$
\Rightarrow \mathbf{m}_j = \frac{\sum_{i=1}^{N} y_{ij} \mathbf{x}_i}{\sum_{i=1}^{N} y_{ij}}
$$

or in simple terms we have:

$$
m_j = \frac{Sum\:of\:all\:dât\:points\:in\: cluster \: j}{Number \: of \: data \: points \: in \: cluster \: j}
$$

```python

def distance(x_1, m_1):
  return np.sqrt(np.sum((x_1 - m_1)**2)) #Defining Euclidean distance
```

```python
def assign_clusters(X, clusters):
    for index_per_x_points in range(X.shape[0]):
        dist = []

        curr_x = X[index_per_x_points] #take every points in X

        for i in range(k): #this would occurs for each points. each X points will be
        #given for each designated centers, with k=3 as of 3 centers
            dis = distance(curr_x, clusters[i]['center']) #find the distance between the current placement
            #and the related data
            dist.append(dis) #and the write down each of the distance inside the dist list
        curr_cluster = np.argmin(dist)#after the iterations of the 3 centers, find the one
        #that have the smallest distance
        clusters[curr_cluster]['points'].append(curr_x) #and then appending the points
        #into the clusters dictionaries.
    return clusters

print(assign_clusters(x, clusters))
```

```python
def update_clusters(X, clusters):
  for i in range(k): #appoint all of the centers.
    points = np.array(clusters[i]['points'])
    if points.shape[0]>0: #so this counts the amount of aggregated points in the
    #point :[] sector, just to make sure if ther eare no points there it could come
    #out as error, halting the progress
      new_center = points.mean(axis=0) # so this finds the new center by find only
      #the mean of the X of the whole points,
      clusters[i]['center'] = new_center #update the new X axis for the randomized center

      clusters[i]['points'] = []
  return clusters


print(update_clusters(x, clusters))



def update_Y_cluster_through_prediction(X,clusters): #this is basically the assign_cluster one
#but this update the Y clusters
  pred = []
  for i in range(X.shape[0]):
    dist = []
    for j in range(k):
      dist.append(distance( X[i], clusters[j]['center'] ))
    pred.append(np.argmin(dist))
  return pred

print(update_Y_cluster_through_prediction(x ,clusters))
```

```python
clusters = assign_clusters(x, clusters)
clusters = update_clusters(x, clusters)
new_y = update_Y_cluster_through_prediction(x ,clusters)

plt.scatter(x[:,0], x[:,1], c=new_y, cmap = ListedColormap(["red", "pink", "green"]))

for i in clusters:
    center = clusters[i]['center']
    plt.scatter(center[0],center[1],marker = '^', c = 'black')
plt.grid(True)
plt.show()
```

##How to determine the optimal K for K-Means

In k-means clustering, the optimal value for 'k' represents the ideal number of clusters that best fit the data. It's the sweet spot where you capture the underlying structure of the data without oversimplifying or overcomplicating the clustering results. Finding the optimal 'k' is crucial for meaningful insights and effective analysis.

###Elbow Methods

Plot the within-cluster sum of squares (WCSS) as a function of the number of clusters. The WCSS measures the squared distance between each point and its assigned cluster center. The idea is to choose the number of clusters at the "elbow" of the plot, where the marginal gain in WCSS decreases significantly.

This is basically the way to find the perfect amount of centers for clustering and further classify stuffs.

```python
from sklearn.cluster import KMeans
wcss = []

iris_df = sns.load_dataset('iris')
iris_df = iris_df.drop(columns=['sepal_length', 'sepal_width'])
iris_df

x = np.array(iris_df.iloc[:,:2])

for k in range(1, 11):
    sklearn_kmeans_ElbowMethods_models = KMeans(n_clusters=k, random_state=42)
    sklearn_kmeans_ElbowMethods_models.fit(X)
    wcss.append(sklearn_kmeans_ElbowMethods_models.inertia_)
'''
Inertia measures how well a dataset was clustered by K-Means.
It is calculated by measuring the distance between each data point and its
centroid, squaring this distance, and summing these squares across one cluster.
A good model is one with low inertia AND a low number of clusters ( K ).
'''
# Plot the WCSS values against the number of clusters (k)
plt.figure(figsize=(9,6))
plt.plot(range(1, 11), wcss)
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

***REMEMBER, THE IDEA IS TO CHOOSE THE NUMBER OF CLUSTERS AT THE "ELBOW" OF THE PLOT***

###The Silhouette Methods

Calculate the average silhouette score for different numbers of clusters. The silhouette score ***measures the similarity of each point to its own cluster compared to other clusters***. The ***optimal number of clusters is the one that maximizes the average silhouette score.***


Note: The silhouette score measures how well each data point fits within its cluster and how far it is from other clusters. It ranges from -1 to 1, where 1 means that the data point is very similar to its own cluster and very dissimilar to other clusters, 0 means that the data point is indifferent or equally similar to all clusters, and -1 means that the data point is more similar to other clusters than its own cluster. A high silhouette score indicates that the clustering algorithm has created well-defined and separated clusters, while a low or negative silhouette score indicates that the clustering algorithm has created overlapping or poorly assigned clusters.


```python
from sklearn.metrics import silhouette_score

# Define range of clusters to evaluate
min_k = 2
max_k = 10

# Initialize list to store silhouette scores
Silhouette_Scores = []

# Loop over different numbers of clusters
for k in range(min_k, max_k+1):
    # Fit K-means to data
    km = KMeans(n_clusters=k, random_state=42)
    labels = km.fit_predict(x)

    # Calculate silhouette score
    Silhouette_Scores.append(silhouette_score(X, labels))

# Plot the silhouette scores against the number of clusters
plt.figure(figsize=(9,6))

plt.plot(range(min_k, max_k+1), Silhouette_Scores)
plt.title('Silhouette Method')
plt.xlabel('Number of clusters')
plt.ylabel('Silhouette score')
plt.show()
```

##advantages and drawbacks visually

K-means clustering is a widely used clustering algorithm that has several advantages and disadvantages:

Advantages:

K-means is a simple, fast, and efficient algorithm that can handle large datasets with high dimensions.
The algorithm is relatively easy to understand and implement, even for non-experts.
The results of K-means clustering are easy to interpret and visualize.
K-means is a popular choice for initialization of other clustering algorithms, such as hierarchical clustering or DBSCAN.
Disadvantages:

K-means is sensitive to the initial choice of cluster centroids (We need to define number of clusters K), which can lead to suboptimal clustering results.
The algorithm assumes that clusters are spherical, equally sized, and have similar densities, which is often not the case in real-world datasets.
K-means can converge to a local minimum and get stuck, especially when the data has noise or outliers.
The algorithm requires the number of clusters to be specified in advance, which can be challenging when the optimal number of clusters is unknown.
Overall, K-means is a useful and widely used clustering algorithm that has its strengths and limitations. It can be a good choice for datasets that meet its assumptions and requirements, but other clustering algorithms may be more suitable for complex or non-spherical datasets.

<h3>15.1.1. Definitions in DBSCAN</h3>

Before learning the DBSCAN algorithm, we define a few key concepts used by it.



<h4>Definition 1:</h4>  
The **epsilon-neighborhood** (also called \( \varepsilon \)-neighborhood) of a data point \( P \) is the set of all points within distance \( \varepsilon \) from \( P \). It is defined as:

$$
N_{\varepsilon}(P) = \{ Q \in D \mid \text{distance}(Q, P) \leq \varepsilon \}
$$

Where $ D $ is the dataset.
And Q is a other random ass datasets

---

<h4>Definition 2:</h4>  
A point \( Q \) is **directly density-reachable** from point \( P \) if:

1.  
$$
Q \in N_{\varepsilon}(P)
$$

2.  
$$
|N_{\varepsilon}(P)| \geq \text{minPts}
$$

This means point $ Q $ is within distance $ \varepsilon $ from $ P $, and point $ P $ lies in a dense region with at least `minPts` points in its neighborhood.

---

<h4>Definition 3:</h4>  
A point \( P_b \) is **density-reachable** from point \( P_a \) if there exists a sequence of points:

$$
P_1, P_2, \dots, P_n \in D
$$

such that:

$$
P_1 = P_a,\quad P_n = P_b
$$

and for every \( i \in \{1, 2, \dots, n-1\} \), the following holds:

$$
P_{i+1} \in N_{\varepsilon}(P_i) \quad \text{and} \quad |N_{\varepsilon}(P_i)| \geq \text{minPts}
$$

This chain of directly density-reachable points connects \( P_a \) to \( P_b \).

---

**Note:**  
Density-reachability is **not symmetric**. That means:

$$
P_b \text{ is density-reachable from } P_a \;\;\not\Rightarrow\;\; P_a \text{ is density-reachable from } P_b
$$

##How to determine epsilon and MinPts parameters of DBSCAN clustering

###Elbow Method

 The elbow method is a popular technique for selecting the value of epsilon. This involves plotting the distances between each data point and its k-th nearest neighbor, where k is the number of neighbors. ##4.3.1 Aggomerative Hierarchical Clustering

For Aggormerative Hierarchical Clustering (AHC) we want to start from the bottom up: aggregating the closest points altogether, and take that create line to create an "area" for the next group of data/data points.

We start with each points as its own cluster, then iteratively merge the closest clusters until one we have only one cluster including all of the data points.

so intuitively speaking, here's the steps. We got n data points

- we then compute the distance between pairs of clusters
-merging the two clusters that are closest
- update the distance matrix to reflect the new cluster

- and we can only **stop** when the is only 1 clusters only.

image.png


I took this picture in w3school. so it is factual that for each datapoints has one other points to have the shortest distance. And for each group of 2 shortest distance, we find the nearest points/group that have the shortest distance too. and so on and so on.






the mathematical concept is quite simple so i dicided to the coding first, and i'll take w2school example, and then i'll try the iris data base.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

x = [4, 5, 10, 4, 3, 11, 14 , 6, 10, 12]
y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21]

data = list(zip(x, y))

plt.figure(figsize=(10,10))
linkage_data = linkage(data, metric='euclidean', method = 'ward')
plt.subplot(1,2,1)
dendrogram(linkage_data)

hierarchical_cluster = AgglomerativeClustering(n_clusters=2, metric='euclidean', linkage='ward')
#linkage can be: ‘ward’, ‘complete’, ‘average’, ‘single’
labels = hierarchical_cluster.fit_predict(data)
plt.subplot(1,2,2)
plt.scatter(x, y, c=labels)

plt.show()
```

```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.preprocessing import LabelEncoder

penguin_datasets = sns.load_dataset('penguins') #this is the first time i use this example datasets
display(penguin_datasets)
penguin_datasets= penguin_datasets.dropna().reset_index(drop=True)
penguin_datasets['sex'] = list(map(lambda x: 0 if x=='Female' else 1, penguin_datasets['sex']))
features = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']
x = penguin_datasets[features].to_numpy()
y = penguin_datasets['species'].to_numpy()
display(penguin_datasets)


LDA_model = LinearDiscriminantAnalysis(n_components=2)
X_LDA = LDA_model.fit_transform(x,y)
X_LDA_datasets = pd.DataFrame(X_LDA, columns=['LDA1', 'LDA2'])
X_LDA_datasets[['sex', 'species', 'island']] = penguin_datasets[['sex', 'species', 'island']]
encoder = LabelEncoder()
X_LDA_datasets['species'] = encoder.fit_transform(X_LDA_datasets['species'])
display(X_LDA_datasets)


plt.figure(figsize=(20,8))
plt.subplot(1,3,1)
plt.title("LDA Model with island hue")
sns.scatterplot(data = X_LDA_datasets, x = 'LDA1', y='LDA2', hue= 'island')
plt.subplot(1,3,2)
sns.scatterplot(data = X_LDA_datasets, x = 'LDA1', y='LDA2', hue= 'sex')
plt.title("LDA Model with island sex")
plt.subplot(1,3,3)
sns.scatterplot(data = X_LDA_datasets, x = 'LDA1', y='LDA2', hue= 'species')
plt.title("LDA Model with island species (the most polarized hue)")
plt.tight_layout()
plt.show()


'''AS WE CAN SEE, THE CLOEST Y IS THE SPECIES, so i'll take the species graph to do an example'''

num_rows, num_cols = 1, 4
k=40
fig, ax = plt.subplots(num_rows, num_cols, figsize=(k, 8))#execute the figure and axes as normal


ax[1].set_title(f"Agglomerative Hierarchial Clustering Method's \n Tree-like diagram with a sample of {k}")
linkage_data = linkage(X_LDA_datasets[['LDA1', 'LDA2']].sample(n=k), metric = 'euclidean')
dendrogram(linkage_data, ax=ax[1])

AggClus_model = AgglomerativeClustering(n_clusters=3, metric = 'euclidean', linkage='ward')
y_species_predict = AggClus_model.fit_predict(X_LDA_datasets[['LDA1', 'LDA2']])
ax[2].set_title("LDA Model with Agglomerative Hierarchial Clustering Method")
sns.scatterplot(ax=ax[2], data = X_LDA_datasets, x = 'LDA1', y='LDA2', c=y_species_predict)

fig.delaxes(ax[0])
fig.delaxes(ax[3])
plt.tight_layout()
plt.show()
```
