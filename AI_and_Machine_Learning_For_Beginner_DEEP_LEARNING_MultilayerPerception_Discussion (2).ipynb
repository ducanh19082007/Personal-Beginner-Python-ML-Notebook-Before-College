##Relations to Gradient Descent

<h2> Understanding the Gradient ∇C in Backpropagation</h2>

$$
\nabla C =
\begin{bmatrix}
\frac{\partial C_o}{\partial w^{(1)}} \\
\frac{\partial C_o}{\partial b^{(1)}} \\
\vdots \\
\frac{\partial C_o}{\partial w^{(L)}} \\
\frac{\partial C_o}{\partial b^{(L)}}
\end{bmatrix}
$$

This is the <strong>gradient vector</strong> of the cost function $C$, where:
- $w^{(l)}$: weights of layer $l$
- $b^{(l)}$: biases of layer $l$
- $L$: the total number of layers in the neural network

---

<h3> How does all of these equations relate to gradient descent?</h3>

Gradient descent uses these gradients to <strong>update the weights and biases</strong> to reduce the cost function:

$$
w^{(l)} := w^{(l)} - \eta \cdot \frac{\partial C_o}{\partial w^{(l)}}
$$

$$
b^{(l)} := b^{(l)} - \eta \cdot \frac{\partial C_o}{\partial b^{(l)}}
$$

Where $\eta$ is the <strong>learning rate</strong>.

So, the vector $\nabla C$ is like a <strong>map of directions</strong>: it tells you how to move every single parameter (weights and biases across all layers) in order to <strong>descend</strong> the loss surface.

```python
def one_hot_encode(y): #just a simple prepreocession process, although i can probably use other methods or Label Encoder but im too lazy
    y_one_hot = []
    y_index = []

    for label in y:
        if label == 'setosa':
            y_one_hot.append([1, 0, 0])
            y_index.append(0)
        elif label == 'versicolor':
            y_one_hot.append([0, 1, 0])
            y_index.append(1)
        elif label == 'virginica':
            y_one_hot.append([0, 0, 1])
            y_index.append(2)

    return np.array(y_one_hot), np.array(y_index)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense #this dictates the type of functions we will use
from tensorflow.keras.layers import Input #like .fit() in sklearn

def create_simple_model():
    model = Sequential([ #Sequential is a linear stack of layers — one layer after another.
                        #Use it when your model has a single input and output and layers flow in a straight line.
        Input(shape=(len_features,)),  # Use Input() as the first layer, and len_features
                                       # is thenumber of dimensions in the input data
        Dense(units=n_neuron_units, activation="relu"), #number of neurons
                                                        #(or units) in the first Dense layer
        Dense(units=n_classes, activation="softmax"), # number of classes to be predicted
        #softmax regression will be mentioned in the next page??
    ])
    model.summary()
    return model

#simple explainantions of what is going on?

'''
This one will be explain in really REALLY simple terms

all of this means that this create a neuron model containing the the certain
input that is relevant to the shape = in the input, and then the second hidden
layer to be dense with n_neuron_units number of neurons, and the second layer
will be executed by ReLU. and the output, or the second layer, to be n_class
number of neurons, executed by softmax regression.
'''




#this will be a detailed explainations



'''
first of all, about the libraries

Sequential: A linear stack of layers — like building blocks stacked one on top
of another. Best for models where the data flows one way only (no loops, no branching).
So basically, this is a way to call out the neural network model and get them into
a variable called "model"

Dense: A fully connected layer. Every neuron in one layer connects to every
neuron in the next. This is the core of most neural networks.

Input: Specifies the shape of your input data. It's like saying
"Hey, my data will have this many features!"


You create a Sequential model (as explained).

The first layer is the Input layer:

shape=(len_features,): This tells the model the input will be a vector with len_features number of values (e.g., if you have 10 features per training example, len_features = 10).



for:  Dense(units=n_neuron_units, activation="relu")
This is the first hidden layer (after input).

units=n_neuron_units: The number of neurons in this layer (you choose how many based on experiment or intuition).

activation="relu": ReLU (Rectified Linear Unit) is a popular non-linear activation function. It helps the model learn non-linear patterns.

for: Dense(units=n_classes, activation="softmax"),
This is the output layer.

units=n_classes: The number of output neurons = number of classes you're predicting (e.g., 3 if you're classifying cats, dogs, and rabbits).

activation="softmax": Converts the output into probabilities that sum to 1. This is used for multiclass classification problems.


'''
```

```python
model = create_simple_model()
```

```python
categorical_cross_entropy_loss = CategoricalCrossentropy()
# CategoricalCrossentropy is chosen as the loss function to measure the difference
# between the model's predictions and the actual labels during training.
#or basically it's like sklearn.metrics.
'''
in fact, this is the substitute for the cost functions that i have discussed before

since (i think) In machine learning / neural networks, the term "cost function"
and "loss function" are often used interchangeably (though there’s a subtle
technical difference in some contexts).

So in this case:
we are using CategoricalCrossentropy() or that $H(p,q) = p(x)log(q(x))$ as the
loss function (a type of cost function) to measure how far the model's
predictions are from the true labels.

It’s especially used when you're doing multi-class classification and my labels
are one-hot encoded


for MNIST database i give an example earlier, using the cost function that i given
could be more effective
'''

# Init optimizer
sgd_opt = SGD(learning_rate=learning_rate)
# SGD is used as the optimizer for the model, and the learning rate is set using the variable `lr`.

'''
this has been discussed continously up there, so i bet you can grasps a bit of what is
going on here

but i just want to recap really quick, so basically, the principal of "fixing"
the weight and the bias is based on the principal of gradient descent, which is
showed and executed here: stochratic gradient descent (SGD). (through hard and trials)

there can be a lot of way to do this, but i choose my opimizer to be sgd though.

like sklearn, this SGD require some stuffs, including:


learning_rate: how big each step is. Which the name describe itself
momentum: to help accelerate gradients in right direction, this is based on the principal
of gradient descent with momentum that i have showed before, increase the percentage of
finding the most minimal value.
nesterov: to enable Nesterov momentum, just a True/False execution
'''

# Compile model
model.compile(loss=categorical_cross_entropy_loss, optimizer=sgd_opt)
# model.compile() is used to configure/to arrange something or put its parts
#together in a particular form or arrangement the model before starting the training process.

'''
just thinks this line is used to ditate all the decision back there, which function
to be used to account for the loss/cost, in order it to know.

And which types of gradient descent to be used. or any kind of mathematical technique
that i probably have not discussed or worse... have not knew.

i should, in fact, reiterate all of the thing here for more detail

the parameter:

loss= means the model/functions used to count the loss of the result for further
assessment of performance, thus using this to fix the model, and imporvements are
indicated by the decreasing of this value each epoch/reiteration

the opimizer= is the the way of fixing stuffs. just as explained before. after the
calculation of finding the ssensitivity of weights and bias, we use this opimizer function
as the way for the neural network's improvements.
'''
```

####ADAM theory

First of all, ADAM stands for Adaptive Moment Estimation

It’s an optimization algorithm that adjusts the model's weights by using gradients — just like Stochastic Gradient Descent (SGD) — but with two powerful improvements, which is:

1. adapts the learning rate for each parameter

2. and uses momentum to smooth out updates

and based on what i ask on Chat GPT, it says: "Adam is basically SGD + Momentum + RMSProp, all in one." which... i think is pretty valid

Mathematically speaking

adam uses the first and second moments of the gradient:
let:

- $g_t$ gradient at time step $t$ -> The derivative of the loss function with respect to the weights — tells the direction to update weights or:` “In which direction should I move this weight to reduce the error?”`
- $m_t$ first moment estimate (mean of gradient) -> The exponential moving average of past gradients — like momentum
`“Let’s not overreact to a single big gradient — let’s average it with what we’ve seen before.”`
- $v_t$ second moment estimate (uncentered variance of gradients) -> The exponential moving average of the squared gradients — tracks how much the gradients fluctuate
`“If the gradient is bouncing around a lot, we should take smaller steps (reduce the learning rate).”`
- $β_1,β_2$ decay rates (e.g. 0.9 and 0.999) -> Control how much weight you give to recent vs. past gradients (usually close to 1, e.g. 0.9 and 0.999)


let this be the case:

$$
m_0 = 0, v_0 = 0
$$

then we can find $m_1$ or the **First moment estimate** as:
$$
m_t = β_1m_{t-1} + (1-β_1)g_t
$$
we can find $v_1$ or the **Second moment estimate** as:
$$
v_t = β_2v_{t-1} + (1-β_2)g^2_t
$$
where:
$$
β_1 ∈ [0,1)\: (usually\: 0.9)
$$
$$
β_2 ∈ [0,1)\: (usually\: 0.999)
$$

furthermore, At the beginning of training, both $m_t$ and $v_t$ are biased toward 0 (since they are initialized as 0). So we fix this with bias-corrected estimates:

$$
\hat{m}_t = \frac{m_t}{1 - \beta_1^t}, \quad
\hat{v}_t = \frac{v_t}{1 - \beta_2^t}
$$

This removes the bias toward zero in early steps.


Finally, update each parameter $\theta $ as:

$$
\theta_{t+1} = \theta_t - \alpha \cdot \frac{\hat{m}_t}{\sqrt{\hat{v}_t} + \epsilon}
$$

Where:

- $ \alpha $: learning rate (e.g. 0.001)  
- $ \epsilon $: a small number (like $ 10^{-8} $) to prevent division by zero

##Softmax Regression foundation

onevsrest (1).png

For short, here is the function:

$$
a_i = sigmoid(z_i) = sigmoid(w^T_ix)=\frac{e^{z_i}}{\sum_{j=1}^Ce^{z_j}}, ∀i = 1,2,...,C
$$

at this point, we have:

$$
p(Y_k = i|x_k;W)=a_i
$$

softmax_ex.png

The softmax function, often used in the final layer of a neural network model for classification tasks, converts raw output scores — also known as logits — into probabilities

##Softmax Regression -- a more stable function

when one of the $z_i$ is too much, the calculation $e^{z_i}$ can make a phenomenon called overflow, affect tremendously with the softmax output. and the is a way to fix this phenomenon by a more stable softmax regression function:

$$
a_i = \frac{e^{z_i}}{\sum^C_{j=1}e^{z_j}} = \frac{e^{-c}e^{z_i}}{e^{-c}\sum^C_{j=1}e^{z_j}} =  \frac{e^{z_i-c}}{\sum^C_{j=1}e^{z_j - c}}=  \frac{exp(z_i-c)}{\sum^C_{j=1}exp(z_j - c)}
$$

with c is a random number. Nevertheless, this number to stablize the answer is usually chosen as:

$$
c = max_i (z_i) = max_i (w^T_ix)
$$

##Softmax Regression Loss Function --- cross-entropy loss

<H3>One Hot Coding Loss function</H3>

when using Softmax Regression, we have: a = softmax(W^Tx), the loss function will be created to minimize the difference between the guessing value of $a$ and the true stuff as $y$

$$
J(W) = \sum^N_{i=1}||a_i-y_i||^2_2
$$

this is probably not a good decision for calculating loss function

<H3>Cross Entropy Loss Function</H3>

with $p$ and $q$ as 2 distinct probability (like y and a)

$$
H(p,q)= -\sum^C_{i=1}p_ilog(q_i)
$$

In **Logistic Regression**, we have a simple binary classification case.

1. The **true output** of a data point \( \mathbf{x}_i \) is a probability distribution vector:
$
\begin{bmatrix}
y_i,
1 - y_i
\end{bmatrix}
$
where $ y_i \in \{0, 1\} $ is the probability that the input point $ \mathbf{x}_i $ belongs to the first class.

2. The **predicted output** of the data point is:

$
a_i = \text{sigmoid}(\mathbf{w}^T \mathbf{x}_i)
$

This gives the predicted probability of $ \mathbf{x}_i $ being in the first class. The probability of it being in the second class is $ 1 - a_i $.

$$
J(w) = -\sum^N_{i=1}(y_i\:log(a_i)+(1-y_i)\:log(1-a_i))
$$

thus, with softmax regression, with C classes, the loss function between the output and the real output of x_i as:

$$
J(W,x_i,y_i) =  -\sum^C_{i=1}y_{ij}log(a_{ij})
$$

so this is like the total of everything, with N amount of data (x,y), we have


$$
J(W,X,Y) =  -\sum^N_{i=1}\biggl(\:\sum^C_{i=1}y_{ij}log(a_{ij})\:\biggl) = -\sum^N_{i=1}\biggl(\:\sum^C_{i=1}y_{ij}\:log\biggl(\frac{e^{z_i-c}}{\sum^C_{j=1}e^{z_j - c}}\biggl)\:\biggl)
$$

<H3>Optimize Loss Function</H3>
with this:

$$
J(W,X,Y) =\:-\sum^C_{i=1}y_{ij}\:log\biggl(\frac{e^{z_i-c}}{\sum^C_{j=1}e^{z_j - c}}\biggl)\:
$$
we have:
$$
J(W,X,Y) = \:-\sum^C_{i=1}y_{ij}\:log\biggl(\frac{e^{z_i-c}}{\sum^C_{j=1}e^{z_j - c}}\biggl)\:=\:-\sum^C_{i=1}y_{ij}w^T_jx_i + log\biggl(\sum^C_{k=1}e^{w^T_kx_i}\biggl)\:=\:-\sum^C_{i=1}y_{ij}a_i + log\biggl(\sum^C_{k=1}e^{a_i}\biggl)\:
$$

and we find the derivative of this, this is heavily inspired by Gradient Descent

$$
\frac{∂J_i(W)}{∂W}= \:-\sum^C_{i=1}y_{ij}x_i + \frac{e^{z_i-c}}{\sum^C_{j=1}e^{z_j - c}} x_i =x_i(a_{ji}-y_{ji})= e_{ji}x_i \:(where \: e_{ji} = a_{ji}-y_{ji})
$$

with the value of $e_{ji} = a_{ji}-y_{ji}$

thus:

$$
\frac{∂J(W)}{∂W} = \sum^N_{i=1}e_{ji}x_i = XE^T
$$

with $E=A-Y$

interestingly, let say we use SGD, with updating weight of $W$

we have:

$$
W = W + η\frac{∂J_i(W)}{∂W}=W + ηx_ie_{ji}^T=  W + ηx_i(y_i-a_i)^T
$$

for short, with SGD format:

$$
W =  W + ηx_i(y_i-a_i)^T
$$
